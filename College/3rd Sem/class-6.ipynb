```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
# Discharging of Capacitor
def f(t,q):
    dqdt = -(q/RC)
    return dqdt
# given
R = 1 # in maga-ohm
C = 5 # in micro-farad
RC = R*C
q0 = 10 # in column

tp = np.linspace(0,20,100)
sol = odeint(f,q0,tp,tfirst=True)

plt.figure(figsize=(7,4))
plt.plot(tp,sol,c="r")
plt.title("Discharging of Capacitor",size=20)

plt.xlim(-0.5,20.5)
plt.ylim(-0.5,10.5)
plt.xlabel("Time (sec) --->",size=15)
plt.ylabel("Q (coulomb) --->",size=15)
plt.xticks(size=12)
plt.yticks(size=12)

plt.grid(ls=":",lw=1.2)
plt.show()
```

```python
# Charging of Capacitor
def f(t,q):
    dqdt = -q/RC + E/R
    return dqdt
# given
R = 1 # in maga-ohm
C = 5 # in micro-farad
RC = R*C
q0 = 0 # in column
E = 10 # in volt

tp = np.linspace(0,20,100)
sol = odeint(f,q0,tp,tfirst=True)


plt.figure(figsize=(7.5,6))
plt.plot(tp,sol,c="r")
plt.title("Charging of Capacitor",size=20)

plt.xlim(-0.5,20.5)
plt.ylim(-0.5,50.5)
plt.xlabel("Time (sec) --->",size=15)
plt.ylabel("Q (coulomb) --->",size=15)
plt.xticks(size=12)
plt.yticks(size=12)

plt.grid(ls=":",lw=1.2)
plt.show()
```

```python
# Radioactive Decay
def f(t,N,lamda):
    dNdt = -lamda*N
    return dNdt
N0 = 1000 # given

tp = np.linspace(0,40,1000)

lam = [0.1,0.2,0.3]
style = ["--","-.","-"]
col = ["g","m","r"]

for i in range(len(lam)):
    sol = odeint(f,N0,tp,args=(lam[i],),tfirst=True)
    plt.plot(tp,sol,c=col[i],ls=style[i],label="lamda = %3.2f" %lam[i])

plt.title("Radioactive Decay",size=20)
plt.xlim(-0.5,40.5)
plt.ylim(-5,1015)
plt.xlabel("Time (sec) --->",size=15)
plt.ylabel("No of Particles --->",size=15)
plt.xticks(size=12)
plt.yticks(size=12)
plt.legend(fontsize=15)
plt.grid(ls=":",lw=1.2)
plt.show()
```
