## Curve Ploting by Python using matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1,10,10)
y = x*2
z = x*3

plt.figure(figsize=(6,4))
plt.title("This is my first plot in python",size=15)

plt.plot(x,y,"o",c="m",ms=8,ls="-",lw=2,label="y = 2x")
plt.plot(x,z,"h",mfc="none",c="b",ms=8,ls="-.",lw=2,label="y = 3z")

plt.xlabel("x-axis",size=14,c="r")
plt.ylabel("y-axis",size=14,c="r")

plt.xticks(size=12,c="k")
plt.yticks(size=12,c="k")

plt.xlim(0,11)
plt.ylim(0,31)

plt.legend(fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

frequency = [1000,1500,2000,2500,3000,3500,4000,4500,5000,5500,6000,6500,7000,7500,8000,8500,9000]
current100 = [0.31,0.52,0.71,0.98,1.14,1.21,1.18,1.11,1.02,0.94,0.86,0.79,0.73,0.67,0.62,0.57,0.53]
current220 = [0.31,0.50,0.68,0.83,0.91,0.93,0.91,0.86,0.80,0.75,0.70,0.65,0.60,0.56,0.52,0.48,0.45]
current330 = [0.30,0.47,0.61,0.71,0.75,0.75,0.73,0.70,0.66,0.62,0.58,0.54,0.50,0.47,0.44,0.41,0.38]

plt.title("Current vs Frequency graph in series LCR Circuit",size=14)
plt.plot(frequency,current100,"o",ms=4,c="r",label="R = 100 ohm")
plt.plot(frequency,current220,"h",ms=4,c="b",label="R = 220 ohm")
plt.plot(frequency,current330,"s",ms=4,c="y",label="R = 330 ohm")

plt.xticks(size=11)
plt.yticks(size=11)

plt.xlabel("frequency (Hz)--->",size=15,c="m")
plt.ylabel("current (mA)--->",size=15,c="m")

plt.xlim(500.9500)
plt.ylim(0.2,1.3)

plt.legend(fontsize=11)
plt.show()
```
