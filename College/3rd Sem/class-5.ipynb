### Curve Fitting (by numpy matplotlib & scipy)

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

def f(x,a,b):
    z = a*x + b
    return z
x = [1,2,3,4,5]
y = [2.25,3.95,6.35,8.0,9.75]

plt.figure(figsize=(5,2.5))
plt.title("Curve Fitting",size=14)
plt.plot(x,y,"o",c="r",ms=6,label="Dotted Line")

par,err = curve_fit(f,x,y)
a = par[0]
b = par[1]
print("a = %5.3f & b = %5.3f" %(a,b))

x_new = np.linspace(0,5,51)
y_fit = f(x_new,a,b)
plt.plot(x_new,y_fit,"-",c="g",lw=2,label="Fitted Line")

plt.xlabel("X axis",size=12)
plt.ylabel("Y axis",size=12)
plt.xticks(size=12)
plt.yticks(size=12)
plt.xlim(0,5.2)
plt.ylim(0,10.2)

plt.legend(fontsize=10)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

def f(lam,a,b):
    z = a + b/(lam**2)
    return z
lam = np.array([400,500,600,700,800,900,1000,1100,1200])
mu = np.array([1.4823,1.4724,1.4670,1.4638,1.4617,1.4602,1.4592,1.4584,1.4572])

plt.figure(figsize=(6.5,4.5))
plt.title("Curve Fitting",size=15)
plt.plot(lam,mu,"o",c="r",ms=6,label="Data Points")

par,err = curve_fit(f,lam,mu)
a = par[0]
b = par[1]
print("a = %5.3f & b = %5.3f" %(a,b))

lam_new = np.linspace(400,1200,801)
mu_fit = f(lam_new,a,b)
plt.plot(lam_new,mu_fit,c="g",lw=2,label="Fitted Line")

plt.xticks(size=11)
plt.yticks(size=11)
plt.xlabel("X axis",size=15,c="brown")
plt.ylabel("Y axis",size=15,c="brown")
plt.xlim(350,1250)
plt.ylim(1.455,1.485)

plt.legend(fontsize=14)
plt.show()
```

### Solution of Differintial Equation

```python
# solution of N = -lamda X N
# Given, at t=0, N0=1000 and lamda=0.063 
import numpy as np
from scipy.integrate import odeint

lam = 0.063
def f(t,N):
    dNdt = -lam*N
    return dNdt

N0 = 1000  # given
tp = np.linspace(0,2,21)
sol = odeint(f,N0,tp,tfirst=True)

for i in range(len(tp)):
    print("In the time %5.3f the solution is %5.3f" %(tp[i], sol[i]))
```
