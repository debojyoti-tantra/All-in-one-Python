## Curve Fitting (Least Square)

```python
import matplotlib.pyplot as plt
import numpy as np

x = [1,2,3,4,5]
y = [2.25,3.95,6.35,8.0,9.75]
n = len(x)
x = np.array(x)
y = np.array(y)

sx = np.sum(x)
sxx = np.sum(x**2)
sy = np.sum(y)
sxy = np.sum(x*y)

A = [n,sx,sx,sxx]
A = np.array(A)
A = A.reshape(2,2)

B = [sy,sxy]
B = np.array(B)
B = B.reshape(2,1)

sol = np.linalg.solve(A,B)
a = sol[0]
b = sol[1]

print("The intersept is:",a)
print("The slope is:",b)

y_fit = a + b*x

plt.figure(figsize=(10,5))
plt.title("Curve Fitting Graph",size=18)

plt.plot(x,y,"o",c="r",ms=7,label="Data Points")
plt.plot(x,y_fit,"-",c="m",lw=2,label="Fitted Line")
plt.legend(fontsize=13)

plt.xticks(size=13)
plt.yticks(size=13)

plt.xlabel("X axis",size=14,c="b")
plt.ylabel("Y axis",size=14,c="b")

plt.xlim(0.8,5.2)
plt.ylim(1.8,10.2)

plt.show()
```
