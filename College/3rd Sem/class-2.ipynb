#### Inverse and determinant of a matrix

```python
import numpy as np
arr = [1,2,-1,2,1,4,3,3,4]
arr = np.array(arr)
M = arr.reshape(3,3)
print("The Matrix is:\n", M)

invM = np.linalg.inv(M)
print("The Inverse of the matrix is:\n", invM)

detM = round(np.linalg.det(M))
print("The Determinant of the Matrix is:\n", detM)
```

#### Verifying the inverse is correct or not

```python
# if determinant value of,
# multiplication of the matrix and the inverse of this matrix is 1
# then inverse is correct
x = np.linalg.det(np.dot(M,invM))
print(round(x))
```

#### Solution of three equation with three variables by matrix multiplication method

```python
# three equations are (solve for x, y and z)
# x + 2y - z = 1
# 2x + y + 4z = 2
# 3x + 3y + 4z = 1
A = np.array([1,2,-1,2,1,4,3,3,4])
A = A.reshape(3,3)
print("Matrix A created by those equations is:\n", A)
B = np.array([1,2,1])
B = B.reshape(3,1)
print("Matrix B created by those equations is:\n", B)
invA = np.linalg.inv(A)
print("Inverse of the Matrix A is:\n", invA)

X = np.dot(invA,B)  # we use the formula: A.X = B => X = inv(A).B
print("The solution of the equation is\n", X)
```

```python
# direct method / solution by direct method numpy.limalg.solve(A,B)
X = np.linalg.solve(A,B)
print("The solution of the equation is\n", X)
```
