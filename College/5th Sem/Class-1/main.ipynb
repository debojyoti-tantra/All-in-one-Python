# Curve Fitting without any python library

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# given
x = [1,2,3,4,5,6]
y = [7,25,30,45,45,60]
```

```python
x_bar = sum(x)/len(x)
y_bar = sum(y)/len(y)

print(x_bar, y_bar)
```

```python
xi_prime = []
for i in range(len(x)):
    xi_prime.append(x[i] - x_bar)

yi_prime = []
for i in range(len(y)):
    yi_prime.append(y[i] - y_bar)

xi_prime_square = []
for i in range(len(xi_prime)):
    xi_prime_square.append((x[i] - x_bar)**2)

xi_prime_yi_prime = []
for i in range(len(x)):
    xi_prime_yi_prime.append(xi_prime[i] * yi_prime[i])

print(xi_prime, yi_prime, xi_prime_square, xi_prime_yi_prime)
```

```python
m = sum(xi_prime_yi_prime)/sum(xi_prime_square)
c = y_bar - m*x_bar

print(m, c)
```

```python
x_new = np.linspace(0, 6, 100)
y_new = []

for i in range(len(x_new)):
    y_new.append(m * x_new[i] + c)

plt.figure(figsize=(6, 4))
plt.title("Curve Fitting", size=15)
plt.scatter(x, y, c='k')

plt.plot(x_new, y_new, c='m')

plt.xlabel("x axis -->", size=14, c='r')
plt.ylabel("y axis -->", size=14, c='r')

plt.xticks(size=12, c='k')
plt.yticks(size=12, c='k')

plt.xlim(-0.5, 6.5)
plt.ylim(-0.5, 65)

plt.grid(c="lightgray")
plt.show()
```
