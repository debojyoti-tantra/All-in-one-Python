# Curve fitting using numpy and scipy

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt
```

```python
# given
x = [1,2,3,4,5,6]
y = [7,25,30,45,45,60]
```

```python
def f(x, a, b):
    z = a*x + b
    return z
```

```python
par, err = curve_fit(f, x, y)
a = par[0]
b = par[1]
print("slope is: ", a, "y-intercept is: ", b)
```

```python
x_new = np.linspace(1, 6, 100)
y_new = f(x_new, a, b)

plt.plot(x, y, "o")

plt.plot(x_new, y_new)

plt.grid()
plt.show()
```
