## Solving Second Order Differential Equation

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
# Dampt vibration differential equation
def f(t,u):
    y = u[0]
    z = u[1]
    dydt = z
    dzdt = -2*k*z - y*(w0)**2
    return [dydt,dzdt]
w0 = 1  # given
k = 0.1  # given
u0 = [1.0,0]  # boundary condition, displacement=1.0 & velocity=0
tp = np.linspace(0,65,10000)
sol = odeint(f,u0,tp,tfirst=True)

plt.figure(figsize=(6.5,3.8))
plt.title("Dampt Vibration (Displacement-Time Graph)",size=15,c="b")
plt.plot(tp,sol[:,0],c="r")
plt.xlabel("Time(sec) --->",size=18)
plt.ylabel("y(t) --->",size=18)
plt.xticks(size=13)
plt.yticks(size=13)
plt.xlim(-1,66)
plt.grid(ls="--")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
# Dampt Vibration differential equation for different values of k
def f(t,u,ke):
    y = u[0]
    z = u[1]
    dydt = z
    dzdt = -2*ke*z - y*(w0)**2
    return [dydt,dzdt]
w0 = 1  # given
k = [0.1,0.5,1.0]  # given
u0 = [1.0,0]  # boundary condition, displacement=1.0 & velocity=0

tp = np.linspace(0,50,10000)

style = ["-.","--","-"]
col = ["r","m","g"]
plt.figure(figsize=(7,4.5))
plt.title("Dampt Vibration for diff values of k\n(Displacement-Time Graph)")
for i in range(len(k)):
    sol = odeint(f,u0,tp,args=(k[i],),tfirst=True)
    plt.plot(tp,sol[:,0],c=col[i],ls=style[i],lw=1.4,label="k=%3.2f"%k[i])

plt.legend(fontsize=15)
plt.xlabel("Time(sec) --->",size=18)
plt.ylabel("y(t) --->",size=18)
plt.xticks(size=13)
plt.yticks(size=13)
plt.xlim(-1,51)
plt.grid(ls="--")
plt.show()
```
